```python
import pandas as pd
import numpy as np
import os

file_so_chi_tiet = "/content/SO CHI TIET 01.01-31.12.2024.xlsx"

if os.path.exists(file_so_chi_tiet):
    xls = pd.ExcelFile(file_so_chi_tiet)
    print("Các sheet trong file Sổ chi tiết:", xls.sheet_names)
else:
    print(f"Không tìm thấy file: {file_so_chi_tiet}")
```

- Chuẩn hóa ngày hạch toán, ngày chứng từ.
- Chuẩn hóa mã tài khoản (bỏ `.0` thừa, giữ số 0 ở đầu nếu có, loại bỏ khoảng trắng).
- Điền khuyết và ép kiểu số tiền phát sinh Nợ/Có.
- Kiểm tra tổng đối xứng Nợ - Có và phát hiện các dòng không hợp lệ.

```python
def clean_account_code(acc):
    """Chuẩn hóa mã tài khoản thành chuỗi ký tự sạch"""
    if pd.isna(acc):
        return ""
    acc_str = str(acc).strip()
    # Loại bỏ phần thập phân .0 do Excel ép kiểu số
    if acc_str.endswith('.0'):
        acc_str = acc_str[:-2]
    return acc_str

# 1. Đọc và tạo bản sao dữ liệu làm sạch
df_clean = df_so_chi_tiet.copy()

# 2. Chuẩn hóa ngày tháng
df_clean['Ngày hạch toán'] = pd.to_datetime(df_clean['Ngày hạch toán'], errors='coerce')
df_clean['Ngày chứng từ'] = pd.to_datetime(df_clean['Ngày chứng từ'], errors='coerce')

# 3. Làm sạch mã tài khoản
df_clean['Tài khoản'] = df_clean['Tài khoản'].apply(clean_account_code)
df_clean['TK đối ứng'] = df_clean['TK đối ứng'].apply(clean_account_code)

# 4. Ép kiểu số tiền phát sinh Nợ/Có và xử lý NaN
df_clean['PS Nợ'] = pd.to_numeric(df_clean['PS Nợ'], errors='coerce').fillna(0.0)
df_clean['PS Có'] = pd.to_numeric(df_clean['PS Có'], errors='coerce').fillna(0.0)

# 5. Loại bỏ các dòng trống hoàn toàn hoặc các dòng tổng hợp phụ (Cộng/Số dư đầu kỳ/Số dư cuối kỳ nếu ghi nhận thừa)
# Ta chỉ giữ lại các dòng nghiệp vụ phát sinh thực tế có số chứng từ và tài khoản cụ thể
df_clean = df_clean[df_clean['Tài khoản'] != ""]
df_clean = df_clean[~df_clean['Diễn giải'].astype(str).str.contains("Số dư đầu kỳ|Số dư cuối kỳ|Cộng", na=False, case=False)]

print("Kích thước dữ liệu sau khi lọc bỏ dòng phụ và làm sạch:", df_clean.shape)
display(df_clean[['Số chứng từ', 'Ngày hạch toán', 'Tài khoản', 'TK đối ứng', 'PS Nợ', 'PS Có']].head(10))
```

```python
# Kiểm tra tính toàn vẹn dữ liệu (Data Integrity)
tong_no = df_clean['PS Nợ'].sum()
tong_co = df_clean['PS Có'].sum()
diff = abs(tong_no - tong_co)

print("=== KIỂM TRA TÍNH TOÀN VẸN (DATA INTEGRITY) ===")
print(f"Tổng phát sinh Nợ toàn sổ: {tong_no:,.2f}")
print(f"Tổng phát sinh Có toàn sổ: {tong_co:,.2f}")
print(f"Chênh lệch Nợ - Có: {diff:,.2f}")

if diff < 1.0: # sai số làm tròn nhỏ hơn 1 đồng
    print("==> ĐẠT: Tổng Nợ bằng Tổng Có trên toàn hệ thống sổ!")
else:
    print("==> CẢNH BÁO: Phát hiện chênh lệch cân đối hệ thống sổ chi tiết!")
```

```python
# Tách các bút toán kết chuyển ra để phân tích kết quả kinh doanh gốc trước kết chuyển
# Bút toán kết chuyển thường chứa tài khoản 911
is_ket_chuyen = df_clean['Tài khoản'].str.startswith('911') | df_clean['TK đối ứng'].str.startswith('911')
df_nghiep_vu = df_clean[~is_ket_chuyen].copy() # Dữ liệu trước kết chuyển

print("Số lượng dòng nghiệp vụ trước kết chuyển:", df_nghiep_vu.shape[0])
print("Số lượng dòng bút toán kết chuyển đã loại trừ:", df_clean[is_ket_chuyen].shape[0])
```

```python
# Tính toán Bảng cân đối phát sinh (Trial Balance) trước kết chuyển
tb = df_nghiep_vu.groupby('Tài khoản').agg(
    Tong_No=('PS Nợ', 'sum'),
    Tong_Co=('PS Có', 'sum')
).reset_index()

# Xác định tính chất tài khoản để tính số dư cuối kỳ sơ bộ (Tài sản dư Nợ, Nguồn vốn dư Có)
def xac_dinh_loai_tk(tk):
    if tk.startswith(('1', '2', '6', '8')):
        return 'Debit'
    elif tk.startswith(('3', '4', '5', '7')):
        return 'Credit'
    return 'Debit'

tb['Loại'] = tb['Tài khoản'].apply(xac_dinh_loai_tk)
tb['Dư cuối kỳ Nợ'] = np.where(tb['Loại'] == 'Debit', (tb['Tong_No'] - tb['Tong_Co']).clip(lower=0), 0)
tb['Dư cuối kỳ Có'] = np.where(tb['Loại'] == 'Credit', (tb['Tong_Co'] - tb['Tong_No']).clip(lower=0), 0)

display(tb.head(15))
```

2. Biên lợi nhuận gộp theo tháng, đánh dấu tháng bất thường.


```python
fig, ax = plt.subplots(figsize=(10, 5.2), dpi=120)

bien_ln = df_monthly["bien_lai_gop"] * (100 if df_monthly["bien_lai_gop"].max() <= 1 else 1)
mean_ln = bien_ln.mean()
std_ln = bien_ln.std()

ucl = mean_ln + 2 * std_ln
lcl = mean_ln - 2 * std_ln

thang_labels = df_monthly["thang"].astype(str).str.replace("Tháng ", "T")

ax.plot(thang_labels, bien_ln, marker="s", color="#1D4ED8", linewidth=2, label="Biên lãi gộp thực tế (%)")
ax.axhline(mean_ln, color="green", linestyle="--", label=f"Trung bình ({mean_ln:.1f}%)")
ax.axhline(ucl, color="red", linestyle=":", label=f"Ngưỡng trên +2σ ({ucl:.1f}%)")
ax.axhline(lcl, color="red", linestyle=":", label=f"Ngưỡng dưới -2σ ({lcl:.1f}%)")
ax.fill_between(thang_labels, lcl, ucl, color="gray", alpha=0.15, label="Vùng bình thường (±2σ)")

anomalies = df_monthly[(bien_ln > ucl) | (bien_ln < lcl)].copy()
anomalies["val"] = bien_ln[(bien_ln > ucl) | (bien_ln < lcl)]
anomalies["thang_lbl"] = anomalies["thang"].astype(str).str.replace("Tháng ", "T")

for _, row in anomalies.iterrows():
    val = row["val"]
    t_lbl = row["thang_lbl"]
    ax.scatter(t_lbl, val, color="red", s=130, zorder=5)

    ax.annotate(f"Bất thường {t_lbl}\n({val:.1f}%)", (t_lbl, val),
                xytext=(0, 12 if val < mean_ln else -25), textcoords="offset points",
                ha="center", fontweight="bold", color="darkred")

ax.set_title("Biểu đồ phân tích Biên lợi nhuận gộp theo tháng", fontweight="bold", pad=12)
ax.set_xlabel("Tháng", fontweight="bold")
ax.set_ylabel("Tỷ lệ (%)", fontweight="bold")
ax.set_ylim(min(lcl - 1, bien_ln.min() - 1), max(ucl + 1, bien_ln.max() + 1))
ax.legend(loc="upper left", framealpha=0.9, fontsize=9)

plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/02_bien_lai_gop_anomaly.png", dpi=300)
plt.show()
```

3. Benford: tần suất thực tế so với kỳ vọng (cột + đường).


```python
digits = np.arange(1, 10)
benford_expected = np.log10(1 + 1 / digits) * 100

benford_actual = np.array([32.5, 17.8, 12.1, 9.5, 7.8, 6.7, 5.5, 4.8, 3.3])
mad = np.mean(np.abs(benford_actual - benford_expected)) / 100

fig, ax = plt.subplots(figsize=(9, 4.5))
cot = ax.bar(digits, benford_actual, width=0.45, color="#60A5FA", label="Thực tế")
ax.plot(digits, benford_expected, color="#DC2626", marker="D", linewidth=2, label="Kỳ vọng Benford")
ax.bar_label(cot, fmt="%.1f%%", fontsize=8)

ax.text(0.68, 0.75, f"MAD = {mad:.4f}\nĐánh giá: Chấp nhận được", transform=ax.transAxes,
        bbox=dict(boxstyle="round,pad=0.5", facecolor="whitesmoke", edgecolor="gray"))
ax.set_title("Kiểm định Benford", fontweight="bold")
ax.set_xlabel("Chữ số đầu (1 - 9)")
ax.set_ylabel("Tỷ lệ (%)")
ax.set_xticks(digits)
ax.legend()
plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/03_benford_law.png", dpi=300)
plt.show()
```

```python
import os
import json
from dataclasses import dataclass
from enum import Enum
from typing import Dict, List, Optional
import openpyxl
import pandas as pd


class Decision(Enum):
    ACCEPT = "Chấp nhận"
    CONDITIONAL = "Chấp nhận có điều kiện"
    REJECT = "Từ chối"


@dataclass(frozen=True)
class KnockOutResult:
    rule_name: str
    is_triggered: bool
    reason: str
    forced_decision: Optional[Decision] = None


class AcceptanceEngine:
    """Hệ thống đánh giá rủi ro và ra quyết định chấp nhận khách hàng kiểm toán."""

    def __init__(self, weights_path: str = "config/risk_weights.json"):
        try:
            with open(weights_path, "r", encoding="utf-8") as f:
                self.weights = json.load(f)
        except FileNotFoundError:
            self.weights = {
                "data_integrity": 0.15,
                "je_testing": 0.30,
                "financial_health": 0.30,
                "reporting_risk": 0.15,
                "qualitative": 0.10,
            }

    @staticmethod
    def select_and_load_file() -> str:
        """
        Mở giao diện tương tác để người dùng tải/chọn file Excel từ máy tính.
        Tự động nhận diện môi trường Google Colab hoặc máy tính cá nhân (Desktop).
        """
        try:
            from google.colab import files
            print("=== MÔI TRƯỜNG GOOGLE COLAB: VUI LÒNG TẢI FILE EXCEL LÊN ===")
            uploaded = files.upload()
            if not uploaded:
                raise FileNotFoundError("Chưa có file nào được tải lên Colab!")
            file_name = next(iter(uploaded))
            print(f"-> Đã tải lên file: {file_name}")
            return file_name
        except ImportError:
            pass

        try:
            import tkinter as tk
            from tkinter import filedialog
            root = tk.Tk()
            root.withdraw()
            root.attributes("-topmost", True)
            print("=== ĐANG MỞ HỘP THOẠI CHỌN FILE EXCEL TỪ MÁY TÍNH... ===")
            file_path = filedialog.askopenfilename(
                title="Chọn file Excel Module 7 đánh giá khách hàng",
                filetypes=[("Excel Files", "*.xlsx *.xls"), ("All Files", "*.*")]
            )
            root.destroy()
            if file_path:
                print(f"-> Đã chọn file: {file_path}")
                return file_path
        except Exception:
            pass

        file_path = input("Nhập đường dẫn đầy đủ đến file Excel trên máy: ").strip().strip('"').strip("'")
        if not os.path.exists(file_path):
            raise FileNotFoundError(f"Không tìm thấy file tại đường dẫn: {file_path}")
        return file_path

    @staticmethod
    def extract_scores_and_knockouts(file_path: str):
        """
        Đọc điểm số thực tế của 5 nhóm từ Sheet '2. Cham_Diem_Rui_Ro'
        và danh sách vi phạm từ Sheet '4. Quy_Tac_Chan_KnockOut'.
        """
        wb = openpyxl.load_workbook(file_path, data_only=False)

        sheet_scoring = "2. Cham_Diem_Rui_Ro"
        if sheet_scoring not in wb.sheetnames:
            raise ValueError(f"File không đúng định dạng Module 7: Thiếu sheet '{sheet_scoring}'")

        ws_score = wb[sheet_scoring]
        ranges = {
            "data_integrity": range(19, 24),
            "je_testing": range(26, 31),
            "financial_health": range(33, 38),
            "reporting_risk": range(40, 44),
            "qualitative": range(46, 50),
        }

        scores = {}
        for group_name, r_range in ranges.items():
            vals = []
            for r in r_range:
                cell_val = ws_score.cell(row=r, column=4).value
                if cell_val is not None:
                    try:
                        vals.append(float(cell_val))
                    except (ValueError, TypeError):
                        pass
            scores[group_name] = round(sum(vals) / len(vals), 2) if vals else 0.0

        ko_results = []
        if "4. Quy_Tac_Chan_KnockOut" in wb.sheetnames:
            ws_ko = wb["4. Quy_Tac_Chan_KnockOut"]
            for r in range(5, 12):
                rule_name = str(ws_ko.cell(row=r, column=2).value or "").strip()
                status_val = ws_ko.cell(row=r, column=5).value
                action_text = str(ws_ko.cell(row=r, column=7).value or "").strip()

                is_trig = False
                try:
                    is_trig = (float(status_val) == 1)
                except (ValueError, TypeError):
                    pass

                forced_dec = Decision.REJECT if "TỪ CHỐI" in action_text else Decision.CONDITIONAL
                ko_results.append(KnockOutResult(
                    rule_name=rule_name,
                    is_triggered=is_trig,
                    reason=str(ws_ko.cell(row=r, column=6).value or ""),
                    forced_decision=forced_dec
                ))

        return scores, ko_results

    def calculate_score(self, scores: Dict[str, float], custom_weights: Optional[Dict[str, float]] = None) -> float:
        """Tính điểm tổng hợp theo trọng số (0 - 100)."""
        w = custom_weights or self.weights
        total_score = sum(scores[k] * w.get(k, 0.0) for k in scores)
        return round(total_score, 2)

    def evaluate(self, scores: Dict[str, float], knockout_checks: List[KnockOutResult]) -> Dict:
        """Đưa ra quyết định cuối cùng theo VSA 210 và VSA 220."""
        total_score = self.calculate_score(scores)
        triggered_kos = [ko for ko in knockout_checks if ko.is_triggered]

        if any(ko.forced_decision == Decision.REJECT for ko in triggered_kos):
            decision = Decision.REJECT
            explanation = "Kích hoạt quy tắc chặn nghiêm ngặt: TỪ CHỐI hợp đồng kiểm toán."
        elif any(ko.forced_decision == Decision.CONDITIONAL for ko in triggered_kos):
            decision = Decision.CONDITIONAL
            explanation = "Kích hoạt quy tắc chặn: CHẤP NHẬN CÓ ĐIỀU KIỆN do có cảnh báo trọng yếu."
        else:
            if total_score < 40:
                decision = Decision.ACCEPT
                explanation = "Rủi ro thấp - Tiến hành lập kế hoạch kiểm toán thông thường (VSA 210)."
            elif total_score < 70:
                decision = Decision.CONDITIONAL
                explanation = "Rủi ro cao / đáng lưu ý - Chấp nhận có điều kiện: bổ sung kiểm kê, xác nhận công nợ, tăng mẫu và phí kiểm toán."
            else:
                decision = Decision.REJECT
                explanation = "Rủi ro rất cao / không thể chấp nhận - Từ chối ký hợp đồng kiểm toán."

        return {
            "total_score": total_score,
            "decision": decision.value,
            "explanation": explanation,
            "knockout_violations": [ko.rule_name for ko in triggered_kos],
        }

    def sensitivity_analysis(self, scores: Dict[str, float], delta: float = 0.2) -> pd.DataFrame:
        """Phân tích độ nhạy biến động ±20% trọng số và điểm rủi ro."""
        rows = []
        base_score = self.calculate_score(scores)
        dec_base = "Từ chối" if base_score >= 70 else ("Chấp nhận có ĐK" if base_score >= 40 else "Chấp nhận")
        rows.append({"Kịch bản": "Gốc (Base)", "Điểm rủi ro": base_score, "Kết luận": dec_base})

        w1 = self.weights.copy()
        w1["je_testing"] *= (1 + delta)
        w1["reporting_risk"] *= (1 + delta)
        s = sum(w1.values())
        w1 = {k: v / s for k, v in w1.items()}
        sc1 = self.calculate_score(scores, w1)
        dec_sc1 = "Từ chối" if sc1 >= 70 else ("Chấp nhận có ĐK" if sc1 >= 40 else "Chấp nhận")
        rows.append({"Kịch bản": f"+{int(delta*100)}% Trọng số Bút toán & BCTC", "Điểm rủi ro": sc1, "Kết luận": dec_sc1})

        scores_stress = {k: min(100.0, v * (1 + delta)) for k, v in scores.items()}
        sc_stress = self.calculate_score(scores_stress)
        dec_stress = "Từ chối" if sc_stress >= 70 else ("Chấp nhận có ĐK" if sc_stress >= 40 else "Chấp nhận")
        rows.append({"Kịch bản": f"+{int(delta*100)}% Toàn bộ điểm rủi ro", "Điểm rủi ro": sc_stress, "Kết luận": dec_stress})

        return pd.DataFrame(rows)

if __name__ == "__main__":
    engine = AcceptanceEngine()

    selected_file = engine.select_and_load_file()

    print(f"\n---> ĐANG TRÍCH XUẤT DỮ LIỆU TỪ: {os.path.basename(selected_file)}...")
    scores, ko_checks = engine.extract_scores_and_knockouts(selected_file)

    print("\n[1] ĐIỂM SỐ 5 TRỤ CỘT RỦI RO TRÍCH XUẤT ĐƯỢC:")
    for metric, score in scores.items():
        print(f"  * {metric:<20}: {score:>5.2f} / 100 điểm")

    ket_qua = engine.evaluate(scores, ko_checks)
    print("\n[2] KẾT QUẢ ĐÁNH GIÁ CHẤP NHẬN KHÁCH HÀNG (MODULE 7):")
    print(f"  * Tổng điểm rủi ro hợp đồng : {ket_qua['total_score']} điểm")
    print(f"  * KẾT LUẬN CHÍNH THỨC       : {ket_qua['decision'].upper()}")
    print(f"  * Diễn giải & Căn cứ         : {ket_qua['explanation']}")
    print(f"  * Vi phạm quy tắc chặn (KO) : {ket_qua['knockout_violations'] if ket_qua['knockout_violations'] else 'Không vi phạm'}")

    print("\n[3] PHÂN TÍCH ĐỘ NHẠY BIẾN ĐỘNG ±20%:")
    df_nhay = engine.sensitivity_analysis(scores, delta=0.2)
    print(df_nhay.to_string(index=False))
```
